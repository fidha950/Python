## **Part A: While Loop** ##

**1.Print the text &quot;Day 1&quot;, &quot;Day 2&quot;, ... up to &quot;Day 7&quot; using a while loop.**

```python
i = 1

while i <= 7:
    print("Day", i)
    i = i + 1
```

**2.The daily sales of a shop for 5 days are stored in a list:**
    
sales = [1200, 1500, 900, 2000, 1800]

```python
sales = [1200, 1500, 900, 2000, 1800]

i = 0

while i < len(sales):
    print("Day", i + 1, "Sales:", sales[i])
    i = i + 1
```

**3.Using the same sales list, find and print the total sales.**

```python
sales = [1200, 1500, 900, 2000, 1800]

i = 0
total = 0

while i < len(sales):
    total = total + sales[i]
    i = i + 1

print("Total Sales:", total)
```

**4.The marks of 5 students are given below. Calculate and print the average marks.**

marks = [45, 78, 62, 90, 55]

```python
marks = [45, 78, 62, 90, 55]

i = 0
total = 0

while i < len(marks):
    total = total + marks[i]
    i = i + 1

average = total / len(marks)

print("Average Marks:", average)
```

**5.Using the same marks list, count how many students scored 50 or more (passed). Print the**

```python
marks = [45, 78, 62, 90, 55]

i = 0
count = 0

while i < len(marks):
    if marks[i] >= 50:
        count = count + 1
    i = i + 1

print("Passed Students:", count)
```

**6.A product costs ₹25. Take the quantity from the user and print the cost for 1 item, 2 items, 3
items ... up to the given quantity. (Example for quantity 3: 25, 50, 75)**

```python
price = 25
quantity = int(input("Enter quantity: "))

i = 1

while i <= quantity:
    print(price * i)
    i = i + 1
```

**7.Find the highest value in the following list using a while loop (do not use max()):**

temperatures = [31, 34, 29, 36, 33]

```python
temperatures = [31, 34, 29, 36, 33]

i = 0
highest = temperatures[0]

while i < len(temperatures):
    if temperatures[i] > highest:
        highest = temperatures[i]
    i = i + 1

print("Highest Temperature:", highest)
```

## **Part B: While with else** ##

**8.Print the month numbers 1 to 3 using a while loop. After the loop finishes, print &quot;Report
Generated&quot; using an else block.**

```python
month = 1

while month <= 3:
    print("Month", month)
    month = month + 1
else:
    print("Report Generated")
```

**9.A list of customer IDs is given:**

ids = [101, 102, 103, 104]

Print each ID using a while loop, and then print &quot;All customers listed&quot; using an else block.

```python
ids = [101, 102, 103, 104]

i = 0

while i < len(ids):
    print(ids[i])
    i = i + 1
else:
    print("All customers listed")
```
