## **Part A: while + break (stop when you find what you need)** ##


**1.The daily sales of a shop for 5 days are given. Find the first day on which the sales were more than 2000. Print the day number and the sales, then stop checking.**

sales = [1200, 1500, 900, 2500, 1800]


```python
sales = [1200, 1500, 900, 2500, 1800]

i = 0

while i < len(sales):
    if sales[i] > 2000:
        print("Day:", i + 1)
        print("Sales:", sales[i])
        break
    i += 1
```

**2.The marks of 5 students are given. Find the first student who failed (marks below 35). Print the student number and the marks, then stop.**

marks = [78, 65, 90, 28, 55]


```python
marks = [78, 65, 90, 28, 55]

i = 0

while i < len(marks):
    if marks[i] < 35:
        print("Student:", i + 1)
        print("Marks:", marks[i])
        break
    i += 1
```

**3.Add the daily sales one by one. Stop as soon as the running total becomes 5000 or more. Print the day number and the total at that point.**

sales = [1200, 1500, 900, 2000, 1800]


```python
sales = [1200, 1500, 900, 2000, 1800]

i = 0
total = 0

while i < len(sales):
    total = total + sales[i]

    if total >= 5000:
        print("Day:", i + 1)
        print("Total:", total)
        break

    i += 1
```

## **Part B: while + continue (skip bad or unwanted values)** ##

**4.A temperature data set has missing values stored as None. Print only the valid temperatures and skip the missing ones.**

temperatures = [31, None, 34, None, 29]


```python
temperatures = [31, None, 34, None, 29]

i = 0

while i < len(temperatures):
    if temperatures[i] is None:
        i += 1
        continue

    print(temperatures[i])
    i += 1
```

**5.Calculate the total of only the positive values in the list. Skip the negative values.**

profit = [500, -200, 800, -100, 300]


```python
profit = [500, -200, 800, -100, 300]

i = 0
total = 0

while i < len(profit):
    if profit[i] < 0:
        i += 1
        continue

    total = total + profit[i]
    i += 1

print("Total profit:", total)
```

**6.The list shows the number of orders placed by 5 customers. Count how many customers placed at least one order. Skip the customers with 0 orders.**

orders = [3, 0, 5, 0, 2]


```python
orders = [3, 0, 5, 0, 2]

i = 0
count = 0

while i < len(orders):
    if orders[i] == 0:
        i += 1
        continue

    count += 1
    i += 1

print("Customers:", count)
```

## **Part C: while + break + continue together** ##

**7.A data list has missing values (None), and the value -1 marks the end of the data. Skip the None values, stop when -1 is found, and print the total of the valid values before it.**

data = [100, None, 250, None, 300, -1, 400]


```python
data = [100, None, 250, None, 300, -1, 400]

i = 0
total = 0

while i < len(data):
    if data[i] is None:
        i += 1
        continue

    if data[i] == -1:
        break

    total = total + data[i]
    i += 1

print("Total:", total)
```

**8.The shop was closed on some days, and the sales for those days are stored as 0. Skip the days with 0 sales and print the other sales. Stop the loop when a sale of more than 2000 is found, and print "Target crossed".**

sales = [1200, 0, 1500, 0, 2500, 1800]


```python
sales = [1200, 0, 1500, 0, 2500, 1800]

i = 0

while i < len(sales):
    if sales[i] == 0:
        i += 1
        continue

    if sales[i] > 2000:
        print("Target crossed")
        break

    print(sales[i])
    i += 1
```

**9.Write a program that keeps asking the user to enter a sale amount and adds it to the total:**

•  If the user enters a negative number, print "Invalid amount" and skip it. (Use continue.)

•  If the user enters 0, stop the program. (Use break.)

•  Otherwise, add the amount to the total.

At the end, print the total sales. (Hint: use while True:)


```python
total = 0

while True:
    amount = float(input("Enter sale amount: "))

    if amount < 0:
        print("Invalid amount")
        continue

    if amount == 0:
        break

    total = total + amount

print("Total sales:", total)
```

## **Part D: while with else + break** ##

**10.A list of customer IDs is given. Take an ID from the user and search for it using a while loop.**

ids = [101, 102, 103, 104]

•  If the ID is found, print its position and stop. (Use break.)

•  If the ID is not found, print "Customer not found" using the else block.


```python
ids = [101, 102, 103, 104]

target = int(input("Enter customer ID: "))

i = 0

while i < len(ids):
    if ids[i] == target:
        print("Position:", i + 1)
        break
    i += 1
else:
    print("Customer not found")
```
