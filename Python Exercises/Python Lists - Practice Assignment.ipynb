### D2. Student Marks ### 

```python
marks = [72, 85, 90, 65, 88, 76, 95, 58]

marks.sort(reverse=True)

print(marks)
print("Top 3:", marks[:3])
print("Bottom 3:", marks[-3:])

total = sum(marks)
average = total / len(marks)

print("Total:", total)
print("Average:", average)
```
