## Set C – Challenge ##

### 1. Record builder ###

```python
record = {}

while True:
    key = input("Enter key: ")

    if key == "stop":
        break

    value = input("Enter value: ")
    record.update({key: value})

print(record)
```

### 2. Word counter ###

```python
sentence = "to be or not to be"

words = sentence.split()
count = {}

for word in words:
    if word in count:
        count.update({word: count[word] + 1})
    else:
        count.update({word: 1})

print(count)
```

### 3. Fail filter ###

```python
marks = {"Asha": 35, "Ravi": 78, "Meena": 29, "John": 60}

for student in list(marks.keys()):
    if marks[student] < 40:
        marks.pop(student)

print(marks)
```

### 4. Cart story ###

```python
cart = {"pen": 10, "book": 50, "bag": 500}

backup = cart.copy()

price = cart.pop("bag")
print(price)

cart.popitem()

cart.update({"pencil": 5})

print(cart)
print(backup)
```

### 5. Shallow copy puzzle ###

```python
a = {"name": "John", "marks": [80, 90]}

b = a.copy()

b["marks"].append(100)
b["name"] = "Sam"

print(a)
```

```python
import copy

a = {"name": "John", "marks": [80, 90]}

b = copy.deepcopy(a)

b["marks"].append(100)
b["name"] = "Sam"

print(a)
print(b)
```
