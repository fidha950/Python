### **Question 8: Unique Pairs Using Frozenset** ###

**A chat app stores conversations between users as pairs. The pair ("Asha", "Bala") and ("Bala", "Asha") are the same conversation.
pairs =** [
    
    ("Asha", "Bala"),
    
    ("Bala", "Asha"),
    
    ("Chitra", "Dinesh"),
    
    ("Asha", "Chitra"),
    
    ("Dinesh", "Chitra"),
    
    ("Asha", "Bala")
]

a) If you put these tuples directly into a set, how many unique pairs do you get? Is that correct for this problem? Why not?

b) Convert each pair into a frozenset and store them in a set. How many unique conversations are there now?

c) Why can't you use a normal set instead of a frozenset inside the outer set here?

d) Create a dictionary that counts how many times each conversation appears, using the frozenset as the key.


```python
pairs = [
    ("Asha", "Bala"),
    ("Bala", "Asha"),
    ("Chitra", "Dinesh"),
    ("Asha", "Chitra"),
    ("Dinesh", "Chitra"),
    ("Asha", "Bala")
]
```

```python
#a) Tuples directly inside a set

unique_pairs = set(pairs)

print(unique_pairs)
print(len(unique_pairs))
```

```python
#b) Using frozenset

unique_conversations = set()

for pair in pairs:
    unique_conversations.add(frozenset(pair))

print(unique_conversations)
print(len(unique_conversations))
```

#### c) Why not normal set? ####

A normal set is mutable and unhashable, so it cannot be placed inside another set.

A frozenset is immutable and hashable, so it can be stored inside an outer set.

```python
#d) Dictionary counting conversations

pairs = [
    ("Asha", "Bala"),
    ("Bala", "Asha"),
    ("Chitra", "Dinesh"),
    ("Asha", "Chitra"),
    ("Dinesh", "Chitra"),
    ("Asha", "Bala")
]

conversation_count = {}

for pair in pairs:
    conversation = frozenset(pair)

    if conversation in conversation_count:
        conversation_count[conversation] += 1
    else:
        conversation_count[conversation] = 1

print(conversation_count)
```
