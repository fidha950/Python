**1. Take a number from the user. Print &quot;Fizz&quot; if it is divisible by 3, &quot;Buzz&quot; if it is divisible by 5,
&quot;FizzBuzz&quot; if it is divisible by both, otherwise print the number itself.**

```python
num = int(input("Enter a number: "))

if num % 3 == 0 and num % 5 == 0:
    print("FizzBuzz")
elif num % 3 == 0:
    print("Fizz")
elif num % 5 == 0:
    print("Buzz")
else:
    print(num)
```

**2. Take the total number of seconds (e.g., 9000) and convert it into hours, minutes and seconds.
(Hint: use // and %.) Print the result like: 2 hours 30 minutes 0 seconds.**

```python
seconds = int(input("Enter total seconds: "))

hours = seconds // 3600
remaining = seconds % 3600
minutes = remaining // 60
seconds = remaining % 60

print(hours, "hours", minutes, "minutes", seconds, "seconds")
```

**3. Take the weight (kg) and height (m) of a person and calculate the BMI (weight / height ** 2).**

    Print the category:

• Below 18.5: Underweight

• 18.5 to 24.9: Normal

• 25 to 29.9: Overweight

• 30 and above: Obese

```python
weight = float(input("Enter weight in kg: "))
height = float(input("Enter height in m: "))

bmi = weight / (height ** 2)

print("BMI:", bmi)

if bmi < 18.5:
    print("Underweight")
elif bmi < 25:
    print("Normal")
elif bmi < 30:
    print("Overweight")
else:
    print("Obese")
```

**4.Take the hour of the day in 24-hour format (0 to 23) and print a greeting:**

• 5 to 11: Good Morning

• Otherwise: Good Night

Print &quot;Invalid time&quot; if the number is outside 0 to 23.

**5. Take a three-digit number and check whether it is a palindrome (e.g., 121, 454). Do not convert
it to a string. (Hint: use // and % to get the digits.)**

```python
num = int(input("Enter a three-digit number: "))

if num >= 100 and num <= 999:
    first = num // 100
    last = num % 10

    if first == last:
        print("Palindrome")
    else:
        print("Not a Palindrome")
else:
    print("Please enter a three-digit number")
```

**6. Take the cost price and selling price of an item. Print whether there is a profit, a loss, or no profit
no loss, along with the amount and the percentage.**

```python
cost = float(input("Enter cost price: "))
selling = float(input("Enter selling price: "))

if selling > cost:
    amount = selling - cost
    percentage = (amount / cost) * 100
    print("Profit:", amount)
    print("Profit Percentage:", percentage)

elif cost > selling:
    amount = cost - selling
    percentage = (amount / cost) * 100
    print("Loss:", amount)
    print("Loss Percentage:", percentage)

else:
    print("No profit no loss")
```

**7. Take the coefficients a, b, c of a quadratic equation ax² + bx + c = 0. Calculate the discriminant (b
** 2 - 4 * a * c) and print whether the roots are real and different, real and equal, or imaginary.**

```python
a = float(input("Enter a: "))
b = float(input("Enter b: "))
c = float(input("Enter c: "))

d = b ** 2 - 4 * a * c

print("Discriminant:", d)

if d > 0:
    print("Roots are real and different")
elif d == 0:
    print("Roots are real and equal")
else:
    print("Roots are imaginary")
```

**10. A taxi charges ₹60 for the first 2 km and ₹15 for every extra km. If the ride is between 10 PM and
5 AM, a night charge of 25% is added to the total fare. Take the distance and the hour (0 to 23)
and print the final fare.**

```python
distance = float(input("Enter distance in km: "))
hour = int(input("Enter hour (0-23): "))

if distance <= 2:
    fare = 60
else:
    fare = 60 + (distance - 2) * 15

if hour >= 22 or hour <= 5:
    fare = fare + (fare * 25 / 100)

print("Final fare: ₹", fare)
```

**11. Take the coordinates (x, y) of a point and print whether it lies on the x-axis, on the y-axis, at the
origin, or in Quadrant 1, 2, 3 or 4.**

```python
x = int(input("Enter x: "))
y = int(input("Enter y: "))

if x == 0 and y == 0:
    print("Origin")
elif y == 0:
    print("X-axis")
elif x == 0:
    print("Y-axis")
elif x > 0 and y > 0:
    print("Quadrant 1")
elif x < 0 and y > 0:
    print("Quadrant 2")
elif x < 0 and y < 0:
    print("Quadrant 3")
else:
    print("Quadrant 4")
```

**12. Swap the values of two variables a and b without using a third variable and without using tuple
