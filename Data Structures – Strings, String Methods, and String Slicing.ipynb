## **3-String Methods** ##

### Upper() ###
Converts all characters into uppercase.


```python
name = "fidha sherin"
print(name.upper())
```

### Lower ###
Converts all characters into lowercase.


```python
name = "FIDHA SHERIN"
print(name.lower())
```

```python
name1 = name.upper()
print(name1)
```

### Title ###
Converts the first letter of each word into uppercase.


```python
text = "data analytics course"
print(text.title())
```

### Capitalize() ###
Converts the first letter of the string into uppercase.


```python
text = "python programming"
print(text.capitalize())
```

### Replace ###
Replaces one value with another.


```python
sentence = "I love Dancing"
print(sentence)
```

```python
print(sentence.replace("Dancing","racing"))
```

### Count() ###
Replaces one value with another.


```python
text = "banana"
print(text.count("a"))
```

### Find ###
Returns the position of a character.


```python
text = "Python"
print(text.find("t"))
```

### Startswith() ###

```python
email = "fidhasherin@gmail.com"
print(email.startswith("fidha"))
```

### Endswith ###

```python
email = "fidhasherin@gmail.com"
print(email.endswith("com"))
```

### Strip ###
Removes extra spaces.


```python
text = " python "
print(text.strip())
```

### Split ###
Splits the string into a list.


```python
text = "python,SQL,Excel"
print(text.split(","))
```

### Join ###
Joins elements into a string.


```python
courses = ["python","SQL","Power BI"]
print("-".join(courses))
```

## **4-String Slicing** ##

What is Slicing?

Slicing is used to extract a part of a string.
    
syntex

string[start:stop:step]


```python
COURSE = "Data Analytics"
print(COURSE[0:4])
print(COURSE[0:5:2])
```

**First Four Character**

```python
text = "Data Analytics"
print(text[0:4])
```

**Characters from Index 4 to End**


```python
print(text[4:])
```

**First Five Characters**


```python
print(text[:5])
```

**Every Second Character**


```python
print(text[::2])
```

**Reverse a String**


```python
print(text[::-1])
```

**Last Three Characters**


```python
print(text[-3:])

```
