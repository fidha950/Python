### **INTRODUCTION TO PYTHON**
Python is a **High Level**, general-purpose programming language.

### **PRINT()**
print() is used to show the result/output of a python program on the screen.

### **VARIABLE**
A variable is a named storage location used to store data in a python program.

### **VARIABLE CREATION**
Variable creation is the process of giving a name to a value so that we can use it later in the program.

### **DATA TYPE**
A data type defines the type or kind of value stored in a variable.

**TYPES**
- int(whole number)
- float(decimal number)
- str(text/string)
- bool(true or false)
- list(collection of value)
- tuple(ordered,unchangeable collection
- set(collection of unique values)
- dict(key-value pairs)

### **1- Integer(int)**
- used to store whole numbers without decimal points.

```python
age = 23
print(age)
```

### **2- Float(float)**
- used to store decimal numbers.

```python
height = 5.4
print(height)
```

### **3- String(str)**
- used to stores text, words, or characters.



```python
name = "Fidha Sherin"
print(name)
```

### **4- Boolean(bool)**
- used to store True or False values.

```python
is_student = True
print(is_student)
```

### **5- List(list)**
- used to store multiple values in an ordered cllection. Values can be changed.

```python
fruits = ["Apple", "Mango", "Orange"]
print(fruits)
```

### **6- Tuple(tuple)**
- used to store multiple ordered values that cannot be changed.

```python
colors = ("Red", "Blue", "Green")
print(colors)
```

### **7- Set(set)**
- used to store unique values. Duplicate values are removed.

```python
numbers = {10, 20, 30, 20}
print(numbers)
```

### **8- Dictionary**
- used to store data in key-values pairs.

```python
student ={"name" : "Fidha Sherin", 
           "age" : 21}
print(student)
```

```python
name = input("Enter your name:")
print(name)
```
