# **Python Lists – List Methods and List Slicing** #

## **Introduction to Lists** ##

### What is a List? ###

A List is a collection of multiple items stored in a single variable.


**Features of Lists:**

Ordered

Mutable (can be modified)
    
Allows duplicate values

Can store different data types


## **Syntax** ##

list_name = [item1, item2, item3]

**Examples**

```python
fruits = ["Apple","Mango","Orange"]
numbers = [10, 20, 30, 40]
mixed = ["Adithya", 25, 75.5, True]

print(fruits)
print(numbers)
print(mixed)
```

### Accessing List Elements ###

```python
print(fruits[0])
print(fruits[1])
print(fruits[2])
```

### Negative Indexing ###

```python
print(fruits[-1])
print(fruits[-2])
print(fruits[-3])
```

## **Part 2: List Methods** ##

### 1. append() ###


Adds an element to the end of the list.


```python
fruits = ["Apple", "Orange"]
print(fruits)
```

```python
fruits.append("Mango")
print(fruits)
```

### 2. insert() ###

Adds an element at a specific position.


```python
fruits = ["Apple", "Mango"]
print(fruits)
```

```python
fruits.insert(1,"Orange")
print(fruits)
```

### 3. extend() ###

Adds multiple elements.


```python
fruits = ["Apple", "Orange"]
print(fruits)
```

```python
fruits.extend(["Mango", "Grapes"])
print(fruits)
```

### 4. remove() ###

Removes a specific value.


```python
fruits = ["Apple", "Orange", "Mango"]
print(fruits)
```

```python
fruits.remove("Orange")
print(fruits)
```

### 5. pop() ###

Removes an element using index.


```python
numbers = [10, 20, 30, 40]
print(numbers)
```

```python
numbers.pop(2)
print(numbers)
```

### Pop Without Index ###

```python
numbers = [10, 20, 30]
print(numbers)
```

```python
numbers.pop()
print(numbers)
```

### 6. clear() ###

Removes all elements.


```python
numbers = [10, 20, 30]
print(numbers)
```

```python
numbers.clear()
print(numbers)
```

### 7. index() ###

Returns the position of an element.


```python
fruits = ["Apple", "Orange", "Mango"]
print(fruits)
```

```python
print(fruits.index("Orange"))
```

### 8. count() ###

Counts occurrences of an element.


```python
numbers = [10, 20, 10, 30, 10]
print(numbers)
```

```python
print(numbers.count(10))
```

### 9. sort() ###

Sorts the list.

```python
numbers = [50, 10, 30, 20]
print(numbers)
```

```python
numbers.sort()
print(numbers)
```

#### Descending Order ####

```python
numbers.sort(reverse=True)
print(numbers)
```

### 10. reverse() ###

Reverses the list.


```python
numbers = [10, 20, 30]
print(numbers)
```

```python
numbers.reverse()
print(numbers)
```

### 11. copy() ###

##### Creates a duplicate/copy of the list. #####

```python
list1 = [10, 20, 30]
print(list)
```

```python
list2 = list1.copy()
print(list2)
```

## **Part 3: List Slicing** ##

### What is List Slicing? ###

List slicing allows us to extract a portion of a list.

### **Syntax** ###

list[start:stop:step] 

### Basic Slicing ###

```python
numbers = [10,20,30,40,50,60]
print(numbers)
```

```python
print(numbers[1:4])
```

### From Beginning ###

```python
numbers = [10,20,30,40,50,60]
print(numbers)
```

```python
print(numbers[:4])
```

### Till End ###

```python
numbers = [10,20,30,40,50,60]
print(numbers)
```

```python
print(numbers[2:])
```

### Entire List ###

```python
numbers = [10,20,30,40]
print(numbers)
```

```python
print(numbers[:])
```

### Using Step Value ###

```python
numbers = [10,20,30,40,50,60]
print(numbers)
```

```python
print(numbers[::2])
```

### Reverse a List Using Slicing ###

```python
numbers = [10,20,30,40,50]
print(numbers[::-1])
```

### Negative Index Slicing ###

```python
numbers = [10,20,30,40,50,60]
print(numbers[-4:-1])
```
